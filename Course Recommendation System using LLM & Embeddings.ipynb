```python
def get_embedding(text):
    response = openai.embeddings.create(
        input=text,
        model='text-embedding-ada-002'
    )
    return response.data[0].embedding
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def get_recommendation_from_user_input(df_embeddings, user_input):
    k=10
    user_embedding = get_embedding(user_input)
    user_embedding = np.array(user_embedding).reshape(1,-1)

    similarities = df_embeddings['embedding'].apply(
        lambda x: cosine_similarity(user_embedding, x.reshape(1,-1))[0][0]
    )

    df_embeddings['similarity'] = similarities

    df_sorted = df_embeddings.sort_values(by='similarity',ascending=False)

    recomendations = []

    for _, row in df_sorted.iloc[0:k].iterrows():
        rec={
            'Title': row['title'],
            'Subject':row['headline'],
            'URL':row['url']
        }
        recomendations.append(rec)
    return recomendations
    
```

```python
get_recommendation_from_user_input(df_embeddings,user_input)
```
